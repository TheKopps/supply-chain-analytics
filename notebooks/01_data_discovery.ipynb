```python
import pandas as pd
from pathlib import Path

DATA_RAW = Path("data/raw")

files = list(DATA_RAW.glob("*.csv"))

for file in files:
    df = pd.read_csv(file)

    print("=" * 80)
    print(f"📁 {file.name}")
    print(f"Shape : {df.shape}")
    print("\nColonnes :")
    print(df.columns.tolist())

    display(df.head())
```

```python
customers = pd.read_csv(DATA_RAW / "olist_customers_dataset.csv")
orders = pd.read_csv(DATA_RAW / "olist_orders_dataset.csv")
order_items = pd.read_csv(DATA_RAW / "olist_order_items_dataset.csv")
payments = pd.read_csv(DATA_RAW / "olist_order_payments_dataset.csv")
products = pd.read_csv(DATA_RAW / "olist_products_dataset.csv")
sellers = pd.read_csv(DATA_RAW / "olist_sellers_dataset.csv")
reviews = pd.read_csv(DATA_RAW / "olist_order_reviews_dataset.csv")
```

```python
datasets = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "payments": payments,
    "products": products,
    "sellers": sellers,
    "reviews": reviews
}

for name, df in datasets.items():
    print(f"{name}: {df.shape}")
```

```python
# Copie de la table principale

sales = orders.copy()
```

```python
sales = sales.merge(
    customers,
    on="customer_id",
    how="left"
)
```

```python
sales = sales.merge(
    order_items,
    on="order_id",
    how="left"
)
```

```python
sales = sales.merge(
    payments,
    on="order_id",
    how="left"
)
```

```python
sales = sales.merge(
    products,
    on="product_id",
    how="left"
)
```

```python
sales = sales.merge(
    sellers,
    on="seller_id",
    how="left",
    suffixes=("", "_seller")
)
```

```python
sales = sales.merge(
    reviews,
    on="order_id",
    how="left"
)
```

```python
print(sales.shape)

sales.head()
```

```python
missing = (
    sales.isnull()
    .sum()
    .sort_values(ascending=False)
)

missing[missing > 0]
```
